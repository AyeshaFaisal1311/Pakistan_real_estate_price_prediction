DATA CLEANING

```python
'''Fill missing values'''
df['agency']=df['agency'].fillna('owner')
df['agent']=df['agent'].fillna('owner')
```

```python
print(df['agency'].isnull().sum())
print(df['agent'].isnull().sum())
```

```python
df.head(5)
```

```python
'''Conversion to one unit of column 'Area' '''
df['area_raw_num']= df['area'].str.extract(r'(\d+)').astype(float)
```

```python
df[['area','area_raw_num']].head(5)
```

```python
df['area_sqft']=np.nan
```

```python
df[['area_sqft']].head(5)
```

```python
#convert Kanal rows
df.loc[df['area'].str.contains('kanal',case=False),'area_sqft'] = df['area_raw_num']* 5445 
#convert Marla rows
df.loc[df['area'].str.contains('marla',case=False),'area_sqft'] = df['area_raw_num']*225
```
